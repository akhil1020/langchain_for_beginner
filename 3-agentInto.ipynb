```python
from langchain.tools import tool

@tool
def get_weather(location:str) -> str:
    """GEt the weather at a location"""
    return f"it's sunny in {location}"
```

### ReAct Architecture

```python
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")

# this built in function create_agent used for make agent very easily
agent=create_agent(
    model=llm,
    tools=[get_weather],
    system_prompt=(
        "You are a helpful assistant. "
        "When you use a tool, you MUST return the tool result directly to the user. "
        "Do not add explanations like 'I don't have real-time data'."
    )
)

# agent
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"what is the weather like in Banglore"}]})
response
```
