### Agent with Memory

```python
import os
from dotenv import load_dotenv
from pprint import pprint

load_dotenv()

OPENAI_API_KEY = os.environ["OPENAI_API_KEY"]
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

## to store current session within the local momory  and we can ruse that context later.

### inmemory saver is used to save the state of the agent in memory. It is useful for testing and debugging purposes. In production, you can use a more persistent storage solution like a database or a file system.
```

```python
agent = create_agent(
    model="gpt-5",
    checkpointer=InMemorySaver()  
    
)
```

```python
### Run with thread id 
config = {"configurable":{"thread_id": "test1"}}
```

```python
from langchain_core.messages import HumanMessage


response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="Hi, my name is Akhil!")
        ]
    },
    config=config
)
response
```

```python
from langchain_core.messages import HumanMessage, system_message


response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="Hi, what is my name ?")
        ]
    },
    config=config
)
response
```
